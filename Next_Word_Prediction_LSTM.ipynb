```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, Dense

rnn_model = Sequential([
    Input(shape=(10,)),

    Embedding(
        input_dim=vocab_size,
        output_dim=100
    ),

    SimpleRNN(128),

    Dense(
        vocab_size,
        activation='softmax'
    )
])

rnn_model.summary()
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, LSTM, Dropout, Dense

lstm_model = Sequential([
    Input(shape=(10,)),

    Embedding(
        input_dim=vocab_size,
        output_dim=100
    ),

    LSTM(128),

    Dropout(0.2),

    Dense(vocab_size, activation='softmax')
])

lstm_model.summary()
```

```python
lstm_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=2,
    restore_best_weights=True
)
```

```python
history_lstm = lstm_model.fit(
    X_train,
    y_train,
    validation_data=(X_test, y_test),
    epochs=10,
    batch_size=64,
    callbacks=[early_stopping]
)
```

```python
def predict_top_words(seed_text, top_n=5):

    token_list = tokenizer.texts_to_sequences(
        [seed_text.lower()]
    )[0]

    token_list = token_list[-10:]

    token_list = pad_sequences(
        [token_list],
        maxlen=10,
        padding='pre'
    )

    # Use the LSTM model
    probabilities = lstm_model.predict(
        token_list,
        verbose=0
    )[0]

    top_indices = np.argsort(
        probabilities
    )[-top_n:][::-1]

    results = []

    for word_id in top_indices:

        word = tokenizer.index_word.get(
            word_id,
            ""
        )

        probability = probabilities[word_id] * 100

        results.append(
            (word, probability)
        )

    return results
```

```python
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences

def predict_next_word(seed_text):

    # Convert text to lowercase
    seed_text = seed_text.lower()

    # Convert text into token IDs
    token_list = tokenizer.texts_to_sequences([seed_text])[0]

    # Keep only the last 10 tokens
    token_list = token_list[-10:]

    # Pad the sequence
    token_list = pad_sequences(
        [token_list],
        maxlen=10,
        padding='pre'
    )

    # Predict the next word
    predicted_probs = lstm_model.predict(
        token_list,
        verbose=0
    )

    # Select the word with the highest probability
    predicted_id = np.argmax(
        predicted_probs,
        axis=-1
    )[0]

    # Convert word ID back to word
    predicted_word = tokenizer.index_word.get(
        predicted_id,
        ""
    )

    return predicted_word
```

```python
def generate_text(seed_text, next_words=5, temperature=0.7):

    for _ in range(next_words):

        # Convert text to token IDs
        token_list = tokenizer.texts_to_sequences([seed_text])[0]

        # Keep last 10 tokens
        token_list = token_list[-10:]

        # Pad sequence
        token_list = pad_sequences(
            [token_list],
            maxlen=10,
            padding='pre'
        )

        # Predict probabilities
        probabilities = lstm_model.predict(
            token_list,
            verbose=0
        )[0]

        # Apply temperature
        probabilities = np.log(
            probabilities + 1e-8
        ) / temperature

        probabilities = np.exp(probabilities)

        # Normalize
        probabilities = probabilities / np.sum(probabilities)

        # Sample next word
        predicted_id = np.random.choice(
            len(probabilities),
            p=probabilities
        )

        predicted_word = tokenizer.index_word.get(
            predicted_id,
            ""
        )

        if predicted_word == "":
            break

        seed_text += " " + predicted_word

    return seed_text
```
